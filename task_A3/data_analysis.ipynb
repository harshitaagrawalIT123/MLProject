# 3. Data Cleaning

## 3.1 Find Number of responses
Expected = 486

```python
import json
from pathlib import Path
from collections import Counter

#json_files should already contain 81 participant json paths
print("JSON files found: ", len(json_files)) # Though, already checked previously

participant_count = 0
response_count = 0

condition_counts = Counter()
shown_counts = Counter()
scenario_counts = Counter()

missing_fields = Counter()
bad_response_counts = []
scenario_mismatches = []
condition_shown_mismatches = []
```

```python
for path in json_files:
    
    # Open a file
    with open(path, "r", encoding = "utf-8") as f:
        data = json.load(f)
    
    # Participant Data - to record participants with != 6 responses
    participant_count += 1
    participant_id = data.get("user_info", []).get("id", path.stem)
    
    # Collect Conditions and Scenarios 
    # (used to count no. of conditions and scenarios)
    user_info = data.get("user_info", {})
    conditions = user_info.get("conditions", [])
    scenarios = user_info.get("scenarios", [])
    responses = data.get("responses", {})
    
    # Record Participants with != 6 responses
    if len(responses) != 6:
        bad_response_counts.append(
            (participant_id, len(responses))
        )
        
    for task_id, response in responses.items():
        
        response_count += 1
        i = int(task_id)
        
        # Condition and scenario assigned to this task
        condition = conditions[i] if i < len(conditions) else None
        assigned_scenario = scenarios[i] if i < len(scenarios) else None
        
        # Information actually recorded in response
        actual_scenario = response.get("scenario")
        model_shown = response.get("model_generation_shown")
        
        condition_counts[condition] += 1
        shown_counts[model_shown] += 1
        scenario_counts[actual_scenario] += 1
        
        # Check Scenario alignment
        if assigned_scenario != actual_scenario:
            scenario_mismatches.append(
                (
                    participant_id,
                    task_id,
                    assigned_scenario,
                    actual_scenario
                )
            )
        
        # Check condition vs model_generation_shown
        if condition in (0, 1) and model_shown in (0, 1):
            if condition != model_shown:
                condition_shown_mismatches.append(
                    (
                        participant_id,
                        task_id,
                        condition,
                        model_shown
                    )
                )

        # Required fields
        for field in [
            "scenario",
            "model_generation",
            "final_version",
            "edits"
        ]:
            if field not in response or response[field] is None:
                missing_fields[field] += 1
```

```python
print("\n========== DATASET SUMMARY ==========")

print("Participants:", participant_count)
print("Response/task records:", response_count)
print("Expected 6 × participants:", participant_count * 6)

print("\nCondition counts:")
for condition, count in sorted(condition_counts.items(), key=lambda x: str(x[0])):
    print(f"  Condition {condition}: {count}")

print("\nmodel_generation_shown counts:")
for value, count in sorted(shown_counts.items(), key=lambda x: str(x[0])):
    print(f"  {value}: {count}")

print("\nScenario counts:")
for scenario, count in sorted(scenario_counts.items(), key=lambda x: str(x[0])):
    print(f"  {scenario}: {count}")

print("\n========== VALIDATION ==========")

print("Participants without exactly 6 responses:",
      len(bad_response_counts))

print("Scenario mismatches:",
      len(scenario_mismatches))

print("Condition/model_shown mismatches:",
      len(condition_shown_mismatches))

print("\nMissing required fields:")
if missing_fields:
    for field, count in missing_fields.items():
        print(f"  {field}: {count}")
else:
    print("  None")

# --------------------------------------------------
# Calculate x
# --------------------------------------------------

x = condition_counts[1]
pure_human = condition_counts[0]

print("\n========== TEXT COUNTS ==========")

print("AI-generated texts stored:", response_count)
print("Human-edited texts (condition 1):", x)
print("Purely human-written texts (condition 0):", pure_human)
print("Total participant final texts:", response_count)

print("\nCheck:")
print(f"{x} + {pure_human} = {x + pure_human}")
```

### 4.2 Create A3 analysis dataframe

```python
records = list()

for path in json_files:
    with open(path, "r", encoding = "utf-8") as f:
        data = json.load(f)
    
    participant_id = data.get("user_info", {}).get("id", {})
    responses = data.get("responses", {})
    
    
    for task_id, response in responses.items():
        if not response.get("scenario"):
            continue
        
        condition = response.get("model_generation_shown")
        
        records.append({
            "participant_id": participant_id,
            "task_id": task_id,
            "scenario": response["scenario"],
            "condition": condition,
            "ai_text": response.get("model_generation"),
            "final_text": response.get("final_version"),
            "edits": response.get("edits", [])
        })
```

```python
df = pd.DataFrame(records)

print("Number of valid task records:", len(df))
print("Number of participants:", df["participant_id"].nunique())

print("\nCondition counts:")
print(df["condition"].value_counts().sort_index())

print("\nScenario counts:")
print(df["scenario"].value_counts().sort_index())

print("\nSum of such counts: ", df["scenario"].value_counts().sort_index().sum())


print("\nMissing values:")
print(df[["ai_text", "final_text"]].isna().sum())

df.head()
```
